** check cosmo code vs CAMB **

```python
from cosmology import cosmo
planck = cosmo()
planck.set_camb_parameters()
planck.normalize()
```

```python
import numpy as np
klist = np.arange(1.0E-4, 1.5, 1.0E-4)
planck.get_nonlin_power()
PlinCAMB = planck.camb_power_lin(klist)
PlinCOSMO = planck.power_spectrumz(klist, z=0)
Pnonlin = planck.camb_power_nonlin(klist)
#Pbbks = planck.power_spectrum_bbks(planck.As, klist)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
ks = 0
plt.loglog(klist[ks:], PlinCOSMO[ks:], "b-", label=r"$P_{\rm lin}^{\rm EH}$")
#plt.loglog(klist[ks:]*0.677, PlinCAMB[ks:], "r-", label=r"$P_{\rm lin}^{\rm CAMB}$")
plt.loglog(klist[ks:], PlinCAMB[ks:], "r:", label=r"$P_{\rm lin}^{\rm CAMB}$")
plt.loglog(klist[ks:], Pnonlin[ks:], "g--", label=r"$P_{\rm nonlin}^{\rm CAMB}$")
#plt.loglog(klist[ks:], Pbbks[ks:], "orange")
plt.legend(loc=0)
plt.xlabel(r'$k\;(h/{\rm Mpc})$')
plt.ylabel(r'$P(k)\; ({\rm Mpc}/h)^3$')
plt.title('CAMB vs EH matter power spectrum test')
```
